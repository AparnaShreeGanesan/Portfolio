8-Merging two list

The program merges two arrays (or lists) by combining their elements into a single list. It adds all elements of the second array to the first array and then prints the merged array.

```python
a = [10, 60, 30, 40]
#declaring flag variable to assume the list is sorted
sorted_flag = True
#loop through the list to compare current element with next element
for i in range(len(a) - 1):
    if a[i] > a[i + 1]:
      sorted_flag = False
      break;    #exit the loop as list is not sorted

if sorted_flag:
  print("List is sorted")
else:
  print("List is not sorted")
```

9-To check if the list is sorted or not

The program checks whether a list is sorted in ascending order. It loops through the list and compares each element with the next one. If any element is greater than the next, the sorted_flag is set to False and the loop stops. Based on the flag, the program prints whether the list is sorted or not

```python
a = [10, 25, 3, 100, 60]
#Assume first element as both largest and second largest
largest = a[0]
second_largest = a[0]
#loop through each element in the list
for i in a:
    if i > largest: #if current element greater than largest
        second_largest = largest #update second largest
        largest = i # current element will become largest
    elif i != largest and i > second_largest: # if the element is between largest and second largest
         second_largest = i #update second largest
print("Second largest element:", second_largest)
```
